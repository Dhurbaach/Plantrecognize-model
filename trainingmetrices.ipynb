```python
#loading finetuned model
finetuned_model = keras.models.load_model('best_model_finetune.keras')
finetuned_model.summary()
```

```python
#Defining the path of the image and loading it
train_dir="plant_dataset/train"
val_dir="plant_dataset/val"
test_dir="plant_dataset/test"
SEED = 42
tf.random.set_seed(SEED)
```

```python
#Loading datasets
from tensorflow.keras.preprocessing import image_dataset_from_directory
IMG_SIZE = (224, 224)
BATCH_SIZE = 32

train_ds = image_dataset_from_directory(train_dir, image_size=IMG_SIZE, batch_size=BATCH_SIZE, label_mode='int', shuffle=True, seed=SEED)
val_ds = image_dataset_from_directory(val_dir, image_size=IMG_SIZE, batch_size=BATCH_SIZE, label_mode='int', shuffle=False)
test_ds = image_dataset_from_directory(test_dir, image_size=IMG_SIZE, batch_size=BATCH_SIZE, label_mode='int', shuffle=False)
```

```python

# Calculate Top-1 and Top-5 accuracy metrics
def calculate_top_k_accuracy(y_true, y_pred_probs, k=5):
    """
    Calculate Top-1 and Top-5 accuracy
    
    Args:
        y_true: Array of true class indices
        y_pred_probs: Model predictions (num_samples, num_classes)
        k: Number of top predictions to consider (default: 5)
    
    Returns:
        top1_acc: Top-1 accuracy percentage
        topk_acc: Top-K accuracy percentage
    """
    top1_correct = 0
    topk_correct = 0
    
    for true_label, probs in zip(y_true, y_pred_probs):
        # Top-1: check if highest probability class matches true label
        top1_pred = np.argmax(probs)
        if top1_pred == true_label:
            top1_correct += 1
        
        # Top-K: check if true label is in top-k predictions
        topk_preds = np.argsort(probs)[-k:]  # Get indices of top-k highest values
        if true_label in topk_preds:
            topk_correct += 1
    
    total = len(y_true)
    top1_accuracy = (top1_correct / total) * 100
    topk_accuracy = (topk_correct / total) * 100
    
    return top1_accuracy, topk_accuracy

# Get predictions on test set
y_true_test = np.concatenate([y for x, y in test_ds], axis=0)
y_pred_probs_test = finetuned_model.predict(test_ds)

# Calculate accuracies
top1_acc, top5_acc = calculate_top_k_accuracy(y_true_test, y_pred_probs_test, k=5)

print(f"\n=== Test Set Accuracy Metrics ===")
print(f"Top-1 Accuracy: {top1_acc:.2f}%")
print(f"Top-5 Accuracy: {top5_acc:.2f}%")
print(f"Improvement from Top-1 to Top-5: {top5_acc - top1_acc:.2f}%")
```
